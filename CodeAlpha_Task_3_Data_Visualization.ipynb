#INTRODUCTION

This project focuses on analyzing a freelancer dataset using Python, Pandas, and data visualization techniques. The purpose of this analysis is to understand freelancer characteristics, earnings, experience, platforms, skills, and other factors.

# Dataset Information

- Total Rows: 6,000
- Total Columns: 36
- Dataset Type: Freelancer Data
- Main Focus: Freelancer profiles, skills, experience, platforms, and monthly earnings

#Data Loading & Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mk=pd.read_csv('/content/drive/MyDrive/pk_freelancing_economy_2015_2025.csv')
```

```python
mk.head()
```

#Data Visualization

```python
mk["platform"].value_counts().plot(
    kind="bar",
    title="Freelancers by Platform",
    xlabel="Platform",
    ylabel="Number of Freelancers"
)
```

```python
mk.groupby("platform")["monthly_earnings_usd"].mean().plot(
    kind="bar",
    title="Average Monthly Earnings by Platform",
    xlabel="Platform",
    ylabel="Monthly Earnings (USD)"
)
```

```python
mk["primary_skill"].value_counts().head(10).plot(
    kind="bar",
    title="Top 10 Primary Skills",
    xlabel="Primary Skill",
    ylabel="Number of Freelancers"
)
```

```python
mk["freelancing_as_primary_income"].value_counts().plot(
    kind="bar",
    title="Freelancing as Primary Income",
    xlabel="Income Status",
    ylabel="Number of Freelancers"
)
```

```python
mk["gender"].value_counts().plot(
    kind="bar",
    title="Gender Distribution",
    xlabel="Gender",
    ylabel="Number of Freelancers"
)
```

```python
mk["education_level"].value_counts().plot(
    kind="bar",
    title="Education Level Distribution",
    xlabel="Education Level",
    ylabel="Number of Freelancers"
)
```

```python
mk["monthly_earnings_usd"].plot(
    kind="hist",
    title="Distribution of Monthly Earnings",
    xlabel="Monthly Earnings (USD)",
    ylabel="Number of Freelancers"
)
```

```python
mk.plot(
    kind="scatter",
    x="jobs_completed",
    y="monthly_earnings_usd",
    title="Jobs Completed vs Monthly Earnings",
    xlabel="Jobs Completed",
    ylabel="Monthly Earnings (USD)"
)
```

```python
mk["experience_group"] = pd.cut(
    mk["experience_years"],
    bins=[0, 1, 3, 5, 10, 11],
    labels=["0-1 years", "1-3 years", "3-5 years", "5-10 years", "10+ years"]
)
```

```python
mk["experience_group"].value_counts().plot(
    kind="bar",
    title="Freelancers by Experience Group",
    xlabel="Experience Group",
    ylabel="Number of Freelancers"
)
```

```python
mk["english_proficiency"].value_counts().plot(
    kind="bar",
    title="English Proficiency Distribution",
    xlabel="English Proficiency",
    ylabel="Number of Freelancers"
)
```

```python
mk["payment_method"].value_counts().plot(
    kind="bar",
    title="Payment Method Distribution",
    xlabel="Payment Method",
    ylabel="Number of Freelancers"
)
```

```python
mk["profile_level"].value_counts().plot(
    kind="bar",
    title="Freelancer Profile Level Distribution",
    xlabel="Profile Level",
    ylabel="Number of Freelancers"
)
```

```python
mk["quarter"].value_counts().plot(
    kind="bar",
    title="Freelancer Records by Quarter",
    xlabel="Quarter",
    ylabel="Number of Freelancers"
)
```
